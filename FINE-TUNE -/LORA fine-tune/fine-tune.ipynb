```python
#### housekeeping
import os
import gc
import torch

# 設定 PyTorch CUDA 記憶體分配行為 # Set PyTorch CUDA memory allocation behavior
os.environ["PYTORCH_CUDA_ALLOC_CONF"] = "expandable_segments:True"

# 釋放 Python 物件占用的 CPU 記憶體 # Release CPU memory occupied by Python objects
gc.collect()

# 如果有 GPU，清空 PyTorch 的 cache
# If a GPU is available, clear PyTorch’s cache
if torch.cuda.is_available():
    torch.cuda.empty_cache()
```

```python

from datasets import Dataset
from transformers import Trainer, TrainingArguments, DataCollatorForLanguageModeling
from transformers import AutoModelForCausalLM
from peft import LoraConfig, get_peft_model, prepare_model_for_kbit_training
import math

OUTPUT_DIR = "./bio_mistral_lora"
SEED = 42
```

```python
# 1️⃣ transform to HuggingFace Dataset
# -------------------------------
from datasets import Dataset
hf_dataset = Dataset.from_list(tokenized)
train_test = hf_dataset.train_test_split(test_size=0.33, seed=SEED)
train_dataset = train_test["train"]
eval_dataset  = train_test["test"]



print("\n=== Train Dataset first 2  ===")
for i in range(2):
    print(train_dataset[i])

print("\n=== Eval Dataset first 2 ===")
for i in range(2):
    print(eval_dataset[i])
```

```python
# 2️⃣ 載入模型並加 LoRA #Load model and apply LoRA
# -------------------------------
model = AutoModelForCausalLM.from_pretrained(
    "BioMistral/BioMistral-7B",
    device_map="auto",
    load_in_8bit=True,
    trust_remote_code=True,
)
model.gradient_checkpointing_enable()
model.config.use_cache = False
model = prepare_model_for_kbit_training(model)

lora_cfg = LoraConfig(
    r=16,
    lora_alpha=32,
    lora_dropout=0.05,
    bias="none",
    target_modules=["q_proj","k_proj","v_proj","o_proj","gate_proj","up_proj","down_proj"],
    task_type="CAUSAL_LM",
)
model = get_peft_model(model, lora_cfg)
model.print_trainable_parameters()
```

```python
# -------------------------------
# 3️⃣ Data collator
# -------------------------------
data_collator = DataCollatorForLanguageModeling(
    tokenizer=tokenizer,
    mlm=False,
    pad_to_multiple_of=8
)

# -------------------------------
# 4️⃣ TrainingArguments
# -------------------------------
training_args = TrainingArguments(
    output_dir=OUTPUT_DIR,
    num_train_epochs=1,
    per_device_train_batch_size=1,
    gradient_accumulation_steps=8,
    learning_rate=2e-4,
    lr_scheduler_type="cosine",
    warmup_ratio=0.03,
    fp16=True,
    logging_steps=10,
    save_steps=20,
    save_total_limit=2,
    gradient_checkpointing=True,
    optim="paged_adamw_32bit",
    seed=SEED,
    report_to="none",
)


# -------------------------------
# 5️⃣ Trainer
# -------------------------------
trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=train_dataset,
    eval_dataset=eval_dataset,
    tokenizer=tokenizer,
    data_collator=data_collator,
)

print("done")
```

```python
# -------------------------------
# 6️⃣ train
# -------------------------------
train_out = trainer.train()
val = trainer.evaluate()

try:
    ppl = math.exp(val["eval_loss"])
    print(f"Validation loss: {val['eval_loss']:.4f} | Perplexity: {ppl:.2f}")
except Exception:
    print(val)


```

```python
# -------------------------------
# 7️⃣ 儲存 LoRA 權重
# -------------------------------
model.save_pretrained(OUTPUT_DIR)
tokenizer.save_pretrained(OUTPUT_DIR)
```

```python
# 參數設定
# -------------------------------
# Parameter settings
MERGED_MODEL = "./mistral7b_subject_merged"
DEVICE = "cuda"  # use GPU
print(2)

# 4-bit 量化設定
# 4-bit quantization settings
bnb_cfg = BitsAndBytesConfig(
    load_in_4bit=True,
    bnb_4bit_quant_type="nf4",           # nf4 / fp4 可選
    # nf4 / fp4 options
    bnb_4bit_compute_dtype=torch.bfloat16,
    bnb_4bit_use_double_quant=True
)

# -------------------------------
# 1️⃣ 載入 tokenizer
# -------------------------------
# 1️⃣ Load tokenizer
tokenizer = AutoTokenizer.from_pretrained(MERGED_MODEL, use_fast=True)

# -------------------------------
# 2️⃣ 載入模型（4-bit 量化 + 自動分配 GPU/CPU）
# -------------------------------
# 2️⃣ Load model (4-bit quantization + automatic GPU/CPU allocation)
model = AutoModelForCausalLM.from_pretrained(
    MERGED_MODEL,
    quantization_config=bnb_cfg,
    device_map="auto"   # 自動將層分配到 GPU/CPU
    # automatically assign layers to GPU/CPU
)

model.eval()  # 設定為推理模式
# set model to evaluation (inference) mode
print("✅ 4-bit merged model loaded successfully")

# -------------------------------
# 3️⃣ 測試生成
# -------------------------------
# 3️⃣ Test generation
prompt = """Given the following symptoms, identify the most likely disease and medical department.

Symptoms: I have trouble breathing and my face and chest are swollen. My eyes are itchy and I have a dry cough."""

inputs = tokenizer(prompt, return_tensors="pt").to(DEVICE)

# 生成
# Generate text
with torch.no_grad():
    output_ids = model.generate(
        **inputs,
        max_new_tokens=64,
        do_sample=True,
        temperature=0.7,
        top_p=0.9
    )

# 轉文字
# Decode to text
print(1)
completion = tokenizer.decode(output_ids[0], skip_special_tokens=True)
print("\n=== Generated Output ===\n")
print(completion)
```
